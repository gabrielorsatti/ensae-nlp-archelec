## 3. Baseline LDA (Full Vocabulary)

We build a document-term matrix using `CountVectorizer` (LDA expects raw counts, not TF-IDF — as discussed in *Chapter VI*) and train a 12-topic LDA model.

### 3.1 Utility function

```python
def get_topic_words(model, feature_names, n_words=15):
    """Extract top words for each topic from a fitted model."""
    topics = {}
    for idx, topic in enumerate(model.components_):
        top_indices = topic.argsort()[-n_words:][::-1]
        topics[idx] = [feature_names[i] for i in top_indices]
    return topics
```

### 3.2 Vectorization and model training

```python
count_vectorizer = CountVectorizer(
    max_features=5000, min_df=10, max_df=0.7,
    stop_words=STOPWORDS_FR, ngram_range=(1, 2)
)
count_matrix = count_vectorizer.fit_transform(df_topic['text_clean'])
feature_names = count_vectorizer.get_feature_names_out()
print(f"Document-term matrix: {count_matrix.shape}")

N_TOPICS = 12

lda = LatentDirichletAllocation(
    n_components=N_TOPICS, max_iter=30,
    learning_method='online', random_state=42, n_jobs=-1
)
lda_topics = lda.fit_transform(count_matrix)

print(f"LDA trained: {N_TOPICS} topics")
print(f"Perplexity: {lda.perplexity(count_matrix):.1f}")
print(f"Log-likelihood: {lda.score(count_matrix):.1f}")
```

### 3.3 Inspect baseline topics

```python
topic_words = get_topic_words(lda, feature_names)
print("=== Baseline LDA Topics ===\n")
for topic_id, words in topic_words.items():
    print(f"Topic {topic_id:2d} : {', '.join(words[:10])}")
```

1. Party names, leader names, and slogans form very distinctive vocabulary clusters

## 7. Model Evaluation & Optimal Number of Topics

### 7.1 Evaluation metrics — theoretical background

As discussed in the course (*Lecture 6 — Topic Modeling: Metrics*), evaluating topic models is challenging because they are **unsupervised** — there is no ground truth to compare against. The course presents three families of metrics:

**Fitting metrics (model-side):**
- **Perplexity** (LDA only): measures how well the model predicts unseen words. Defined as the inverse of the geometric mean per-word likelihood. Lower is better. However, as stated in the slides: *"Fitting metrics such as Perplexity (LDA) or reconstruction error (NMF) are useful to compare models but are not related to human ratings of topic coherence."*
- **Reconstruction error** (NMF only): measures how well $W \times H$ approximates the original matrix. Lower is better.

**Coherence metrics (human-aligned):**
These evaluate whether the top words of each topic are **semantically related** — i.e., whether they tend to co-occur in real text. The course describes three coherence scores:

- **UMass Coherence** ($C_{UMass}$): computed on the **word sequence of the training texts**. Measures how often the top topic words co-occur within the same documents. Higher (less negative) is better. Fast to compute but biased toward the training corpus.

- **UCI Coherence** ($C_{UCI}$): computed on a **sliding window** over training texts (or an external corpus). Based on Pointwise Mutual Information (PMI) between word pairs. Higher is better.

- **$C_v$ Coherence**: the most sophisticated metric, described in the slides as *"a metric discovered through systematic experiments to find the correlation between coherence metrics and human ratings."* It combines NPMI vectors with cosine similarity. **$C_v$ is considered the closest to human judgment** (Röder et al., 2015).

Our evaluation strategy: we compute **perplexity** (for LDA), **reconstruction error** (for NMF), and **$C_v$ coherence** (for both) across a range of $k$ values, then select the optimal $k$ that balances statistical fit with human-interpretable coherence.

### 7.2 Vectorization (clean vocabulary)

```python
count_vec_clean = CountVectorizer(
    max_features=3000, min_df=15, max_df=0.6,
    stop_words=STOPWORDS_PARTISAN, ngram_range=(1, 2),
)
count_matrix_clean = count_vec_clean.fit_transform(df_topic['text_clean'])
feature_names_clean = count_vec_clean.get_feature_names_out()

tfidf_vec_clean = TfidfVectorizer(
    max_features=3000, min_df=15, max_df=0.6,
    stop_words=STOPWORDS_PARTISAN, ngram_range=(1, 2),
)
tfidf_matrix_clean = tfidf_vec_clean.fit_transform(df_topic['text_clean'])
tfidf_features_clean = tfidf_vec_clean.get_feature_names_out()

print(f"Clean Count matrix: {count_matrix_clean.shape}")
print(f"Clean TF-IDF matrix: {tfidf_matrix_clean.shape}")
```

### 7.3 Grid search: LDA evaluation

We train LDA models for $k \in \{5, 7, 8, 10, 12, 15, 20\}$ and compute perplexity, log-likelihood, and $C_v$ coherence for each.

```python
# Install gensim if needed (for coherence scores)
try:
    import gensim
    from gensim.models import CoherenceModel
    from gensim.corpora import Dictionary
    print(f"gensim version: {gensim.__version__}")
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "gensim", "--quiet"])
    import gensim
    from gensim.models import CoherenceModel
    from gensim.corpora import Dictionary
    print(f"gensim installed: {gensim.__version__}")

# Prepare gensim corpus for coherence computation
texts_tokenized = [text.split() for text in df_topic['text_clean']]
gensim_dictionary = Dictionary(texts_tokenized)
gensim_corpus = [gensim_dictionary.doc2bow(text) for text in texts_tokenized]
print(f"Gensim dictionary: {len(gensim_dictionary)} terms")
print(f"Gensim corpus: {len(gensim_corpus)} documents")
```

```python
range_topics = [5, 7, 8, 10, 12, 15, 20]

# LDA evaluation
lda_perplexities = []
lda_log_likelihoods = []
lda_coherences_cv = []
lda_coherences_umass = []

print("=" * 70)
print("LDA Grid Search (clean vocabulary)")
print("=" * 70)

for n in range_topics:
    # Train sklearn LDA
    lda_test = LatentDirichletAllocation(
        n_components=n, max_iter=30, learning_method='online',
        learning_decay=0.7, random_state=42, n_jobs=-1,
        doc_topic_prior=0.1, topic_word_prior=0.01,
    )
    lda_test.fit(count_matrix_clean)
    
    # Fitting metrics
    perp = lda_test.perplexity(count_matrix_clean)
    ll = lda_test.score(count_matrix_clean)
    lda_perplexities.append(perp)
    lda_log_likelihoods.append(ll)
    
    # Extract top words for coherence
    topic_top_words = []
    for topic in lda_test.components_:
        top_indices = topic.argsort()[-10:][::-1]
        topic_top_words.append([feature_names_clean[i] for i in top_indices])
    
    # Cv coherence (closest to human judgment)
    cm_cv = CoherenceModel(
        topics=topic_top_words, texts=texts_tokenized,
        dictionary=gensim_dictionary, coherence='c_v'
    )
    cv_score = cm_cv.get_coherence()
    lda_coherences_cv.append(cv_score)
    
    # UMass coherence
    cm_umass = CoherenceModel(
        topics=topic_top_words, texts=texts_tokenized,
        dictionary=gensim_dictionary, coherence='u_mass'
    )
    umass_score = cm_umass.get_coherence()
    lda_coherences_umass.append(umass_score)
    
    print(f"  k={n:2d}  |  Perplexity: {perp:>10,.1f}  |  Log-LL: {ll:>14,.1f}  |  Cv: {cv_score:.4f}  |  UMass: {umass_score:.4f}")

print("\nDone.")
```

### 7.4 Grid search: NMF evaluation

NMF does not have perplexity (it is not a probabilistic model — see the comparison table in the course slides: *"LDA = probabilistic generative model, NMF = linear algebra factorization"*). Instead, we use **reconstruction error** alongside coherence scores.

```python
# NMF evaluation
nmf_recon_errors = []
nmf_coherences_cv = []
nmf_coherences_umass = []

print("=" * 70)
print("NMF Grid Search (clean vocabulary)")
print("=" * 70)

for n in range_topics:
    nmf_test = NMF(
        n_components=n, random_state=42, max_iter=500, init='nndsvda'
    )
    W = nmf_test.fit_transform(tfidf_matrix_clean)
    
    # Reconstruction error
    recon_error = nmf_test.reconstruction_err_
    nmf_recon_errors.append(recon_error)
    
    # Extract top words for coherence
    topic_top_words = []
    for topic in nmf_test.components_:
        top_indices = topic.argsort()[-10:][::-1]
        topic_top_words.append([tfidf_features_clean[i] for i in top_indices])
    
    # Cv coherence
    cm_cv = CoherenceModel(
        topics=topic_top_words, texts=texts_tokenized,
        dictionary=gensim_dictionary, coherence='c_v'
    )
    cv_score = cm_cv.get_coherence()
    nmf_coherences_cv.append(cv_score)
    
    # UMass coherence
    cm_umass = CoherenceModel(
        topics=topic_top_words, texts=texts_tokenized,
        dictionary=gensim_dictionary, coherence='u_mass'
    )
    umass_score = cm_umass.get_coherence()
    nmf_coherences_umass.append(umass_score)
    
    print(f"  k={n:2d}  |  Recon. Error: {recon_error:>10,.4f}  |  Cv: {cv_score:.4f}  |  UMass: {umass_score:.4f}")

print("\nDone.")
```

### 7.8 Decision: selecting the optimal model and $k$

Based on the evaluation above, we make the following decisions:

**Selection criteria** (following the course hierarchy):
1. **$C_v$ coherence** is our primary metric, as it is the most correlated with human topic ratings (Röder et al., 2015 — cited in the course slides).
2. **UMass coherence** serves as a secondary check (computed directly on the training corpus).
3. **Perplexity / Reconstruction error** are informative but, as the course explicitly warns, *"not related to human ratings of topic coherence."*

We select the $k$ that maximizes $C_v$ coherence. Between LDA and NMF, we compare their best $C_v$ scores and also consider interpretability (as the course emphasizes: *"Topic modelling is unsupervised and requires human interpretation"*).

The improved LDA and NMF models will both be trained at the selected $k$ below, and their topics will be inspected to make a final human judgment.

### 9.7 Most exclusive words per topic

```python
print("=== Most EXCLUSIVE words per topic ===")
print("(words that best distinguish each theme from all others)\n")

for t in range(best_n_topics):
    weights = best_model.components_[t]
    other_weights = np.mean(np.delete(best_model.components_, t, axis=0), axis=0)
    with np.errstate(divide='ignore', invalid='ignore'):
        exclusivity = np.where(other_weights > 0, weights / other_weights, 0)
    mask_min_weight = weights > np.percentile(weights, 90)
    exclusivity_filtered = exclusivity * mask_min_weight
    top_exclusive = exclusivity_filtered.argsort()[-8:][::-1]
    words = [best_features[i] for i in top_exclusive]
    topic_name = TOPIC_NAMES_CLEAN.get(t, f"Topic {t}")
    print(f"  Topic {t:2d} ({topic_name:45s}) : {', '.join(words)}")
```
